```r
# histogram plot for question 1: numeric variable

options(repr.plot.width = 8, repr.plot.height = 8) 


bb_t_hist <- ggplot(bb_tidy_data, aes(x = total_wins_pct)) +
                geom_histogram(bins = 30) +
                labs(x = "Total Win Percentage \n (Out of 100)", y = "Number of Percentage Values", 
                     title = "A Plot of the Distribution of \n Total Win Percentage Values") +
                theme(text = element_text(size = 14))
                
bb_t_hist
```
